In this project, we modeled heartbeats using RR interval sequences from the MIT-BIH Arrhythmia Database and generated synthetic rhythms using a Markov transition model. We successfully captured the overall distribution of the RR intervals and the short-term transition behavior between slow, normal, and fast heart rates. For mainly normal rhythm, which was Record 100, the synthetic sequences shared variability and shape with the original sequences. However, for the arrhythmic rhythm, which is record 208, the synthetic sequences lacked the multi-beat irregular patterns and pauses of the real sequence. 

The limitations arise from the structure of the model. For example, the Markov chain (first-order) only looks at one beat at a time. However, real arrythmias doesn't work like this. It's mainly occurring across patterns that stretch across many consecutive beats. Since this is a longer-term structure, our first-order model can't capture this, as it depends solely on the previous state, and not the ones before that. This is why our synthetic arrhythmia looks a lot more random than the real data.

For our future work, adding a higher-order Markov chain would be a good idea, since we could go back more than one state. This would help the model capture perhaps 3 or 4 beats prior when predicting the next one, which seems to be important as we see in the real data.
